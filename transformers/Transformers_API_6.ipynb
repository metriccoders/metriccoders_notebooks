```python
question="Who is Rahul Dravid?"
context = "Rahul Dravid is one of India's greatest test cricketers"
```

```python
from transformers import pipeline
```

```python
ques_ans = pipeline("question-answering", model="deepset/roberta-base-squad2")
```

```python
ques_ans(question=question, context=context)
```
